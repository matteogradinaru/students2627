#### Our first step: splitting our data

Splitting your dataset correctly is the first critical step in a machine learning project. We partition the data into two distinct sets: one for training and another for testing. By training our model on the first set and evaluating it on the second, we aim to create a model that generalizes. A model that **generalizes** has effectively learned the underlying patterns in the dataset rather than just memorizing the training data. This skill is essential to avoid **overfitting**, where a model might excel with the training data but fail to predict accurately on new, unseen data. 

Throughout this course, we'll focus on how to achieve generalization to ensure that our models are robust and perform well in practical applications.

💻📊💡 sci-kit learn typically uses a common interface across all its functionality:

1.  `Fit` means "prepare" this object. For instance, we `fit` a machine learning model, that means its trained and we can use it later. We also `fit` a one hot encoder. We prepare it on existing data. For our `OneHotEncoder` fit essentially does two things: it finds all existing columns and assigns them a number.
2.  `transform` means to apply a preprocessing transformation. Typically you need to have `fit` it first. 
3.  `fit_transform` applies a `fit` and a `transform` in one go.
4.  `predict` makes a prediction with a machine learning model.
5.  `fit_predict` is analogous to `fit_transform`, it trains a machine learning model and then instantly makes a prediction on the same data.

##### ❓ Interpret the meaning of the numbers in the one hot encoding output you see

My answer:
The output consists entirely of 0s and 1s. A 1 indicates the presence of that specific category for that specific row, while a 0 indicates its absence. Each unique category in the original column becomes its own new binary column.

```python
cat_columns = ["package_Type", "destination", "country"]
```

```python
ohe = OneHotEncoder(sparse_output=True)
ohe.fit(X_train[cat_columns])
cat_cols_train  = ohe.transform(X_train[cat_columns])
cat_cols_train
```

```python
# For the rest of the demo we need the non-sparse format
ohe = OneHotEncoder(sparse_output=False)
ohe.fit(X_train[cat_columns])
cat_cols_train  = ohe.transform(X_train[cat_columns])
```

##### ❓ Don't be afraid of the output above. What does it mean? How do you interpret it? What's dthe difference between sparse and non-sparse matrices? 

My answer:
A non-sparse (dense) matrix stores every single 0 and 1 in your computer's memory. A sparse matrix optimizes this by only storing the coordinates/positions of the non-zero elements (the 1s). Because one-hot encoding generates mostly zeros, sparse matrices save massive amounts of RAM.

##### ❓ HARDER: can you see the dangerous thing about one hot encoding?

My answer:
The "Curse of Dimensionality." If a categorical column has hundreds of unique values (high cardinality), OHE will create hundreds of new columns. This bloats the dataset, drastically slows down training time, increases memory consumption, and makes the model highly susceptible to overfitting.

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
import numpy as np

# 1. Automatically detect text vs numeric columns to bypass spelling/capitalization errors
cat_columns = X_train.select_dtypes(include=['object', 'bool', 'category']).columns.tolist()
numeric_columns = X_train.select_dtypes(include=['number']).columns.tolist()

# (Optional safety net: remove ID columns if they exist so they don't mess up the math)
cat_columns = [col for col in cat_columns if 'id' not in col.lower()]
numeric_columns = [col for col in numeric_columns if 'id' not in col.lower()]

# 2. Initialize the Scikit-Learn tools
ohe = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
scaler = StandardScaler()

# 3. Preprocess the TRAINING data (fit and transform)
cat_cols_train = ohe.fit_transform(X_train[cat_columns])
num_cols_train = scaler.fit_transform(X_train[numeric_columns])
X_train_preprocessed = np.hstack((cat_cols_train, num_cols_train))

# 4. Preprocess the TEST data (transform ONLY)
cat_cols_test = ohe.transform(X_test[cat_columns])
num_cols_test = scaler.transform(X_test[numeric_columns])
X_test_preprocessed = np.hstack((cat_cols_test, num_cols_test))
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
import pandas as pd
import plotly.express as px

# 1. Initialize and train BOTH models on the fresh 33-feature data
linear_model = LinearRegression()
linear_model.fit(X_train_preprocessed, y_train)

knn_model = KNeighborsRegressor()
knn_model.fit(X_train_preprocessed, y_train)

# 2. Predict on the test set using both models
lin_test_preds = linear_model.predict(X_test_preprocessed)
knn_test_preds = knn_model.predict(X_test_preprocessed)

# 3. Calculate the errors (Actual - Predicted)
lin_errors = y_test - lin_test_preds
knn_errors = y_test - knn_test_preds

# 4. Combine into a dataframe for easy plotting
error_df = pd.DataFrame({
    'Linear Regression Errors': lin_errors,
    'KNN Errors': knn_errors
})

# 5. Visualize the distribution of the errors
fig_errors = px.histogram(
    error_df, 
    title="Distribution of Prediction Errors on Test Set (Actual Cost - Predicted Cost)",
    labels={'value': 'Error Amount', 'variable': 'Model'},
    barmode='overlay',
    opacity=0.7
)
fig_errors.show()
```

My Answer/Observation:
An error of 0 means a perfect prediction. The model with the tallest spike closest to 0 on the x-axis is making the most accurate predictions. While KNN likely looked much better on the training data, Linear Regression might actually perform better (or similarly) on the test data. If KNN's errors are more spread out on this test set, it indicates the KNN model overfit the training data and failed to generalize as well as the simpler Linear Regression model.
